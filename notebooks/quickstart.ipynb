# Quickstart

````python
from IPython.display import Markdown, HTML
import geojson
import folium
import geomd as gmd
````

## GeoMarkdown to GeoJSON

### Parse and render

````python
geomd_str = """

# Barcelona Tour

## Park Güell

<!--
field_mapping:
    - On map: geometry
-->

### Overview

Park Güell is a public park system composed of gardens and architectural elements located on Carmel Hill

### Address

08024 Barcelona, Spain

### On map

```geometry
{
    "type": "Point",
    "coordinates": [2.152777, 41.414444]
}
```

## Casa Batlló

<!--
field_mapping:
    - On map: geometry
-->

### Overview

Casa Batlló is a renowned building on Passeig de Gràcia, remodeled by Antoni Gaudí between 1904 and 1906

### Address

Passeig de Gràcia, 43, 08007 Barcelona, Spain

### On map

```geometry
{
    "type": "Point",
    "coordinates": [2.16487, 41.391642]
}
```
"""
````

````python
Markdown(
    f'<div style="background-color: #E5FFCC; padding: 24px">{geomd_str}</div>'
)
````

````python
geojson_str = gmd.convert(
    text=geomd_str,
    src="geomd",
    dst="geojson"
)
geojson_str
````

````python
def display_on_map(geojson_data):
    # Create a map instance
    m = folium.Map()
    
    # Create a Tooltip
    tooltip = folium.GeoJsonTooltip(
        fields=["title"],
        aliases=["Name:"],
        # labels=True,
        # sticky=True
    )
    
    # Create a Popup
    popup = folium.GeoJsonPopup(
        fields=["title", "Overview", "Address"],
        aliases=["Name:", "Overview:", "Address:"],
        localize=True
    )
    
    # Add your GeoJSON data
    layer = folium.GeoJson(
        geojson_data,
        name=geojson_data['title'],
        tooltip=tooltip,
        popup=popup
    ).add_to(m)

    # Fit a map content
    m.fit_bounds(layer.get_bounds())
    
    display(m)
````

````python
geojson_data = geojson.loads(geojson_str)
display_on_map(geojson_data)
````

## GeoJSON to GeoMarkdown

````python
geojson_str = """
{
    "type": "FeatureCollection", 
    "title": "Barcelona Tour", 
    "features": [
        {
            "type": "Feature", 
            "title": "Park Güell", 
            "geometry": {"type": "Point", "coordinates": [2.152777, 41.414444]}, 
            "properties": {
                "id": "1", 
                "author": "Adam Smith", 
                "overview": "Park Güell is a public park system composed of gardens and architectural elements located on Carmel Hill", 
                "address": "08024 Barcelona, Spain", 
                "title": "Park Güell"
            }
        }, 
        {
            "type": "Feature", 
            "title": "Casa Batlló", 
            "geometry": {"type": "Point", "coordinates": [2.16487, 41.391642]}, 
            "properties": {
                "id": "2", 
                "author": "Adam Smith", 
                "overview": "Casa Batlló is a renowned building on Passeig de Gràcia, remodeled by Antoni Gaudí between 1904 and 1906", 
                "address": "Passeig de Gràcia, 43, 08007 Barcelona, Spain", 
                "title": "Casa Batlló"
            }
        }
    ]
}
"""
````
